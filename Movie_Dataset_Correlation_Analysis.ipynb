```python
import pandas as pd
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt
import matplotlib.mlab as mlab
import matplotlib
```

```python
plt.style.use('ggplot')
from matplotlib.pyplot import figure

%matplotlib inline
matplotlib.rcParams['figure.figsize'] = (12,8)
```

```python
# read in the data
df = pd.read_csv(r'C:\Users\ben71\Downloads\movies\movies.csv')
```

```python
#looking at the data

df.head()
```

```python
#seeing if there is missing data
for col in df.columns:
    pct_missing = np.mean(df[col].isnull())
    print('{} - {}%'.format(col,pct_missing))
```

```python
# data types for our columns
df.dtypes
```

```python
#dropping all the null values in the tables
df = df.dropna(subset=['budget'])
df = df.dropna(subset=['gross'])

#changing datatype of the columns
df['budget'] = df['budget'].astype('int64')
df['gross'] = df['gross'].astype('int64')
```

```python
df
```

```python
#creating correct year column
#str.rsplit(' (', n=-1) splits the text where the find the open bracket. str[0] takes only the date part out of the two items that were split. str.split(',', n=1).str[-1] splits the date and year and takes only the year value
df['year_corrected'] = df['released'].str.rsplit(' (', n=-1).str[0].str.split(',', n=1).str[-1]
```

```python
df['year_corrected']
```

```python
df = df.sort_values(by=['gross'], inplace = False, ascending = False)
```

```python
pd.set_option('display.max_rows',None)
```

```python
# dropping any duplicates
df['company'].drop_duplicates().sort_values(ascending = False)
```

```python
df
```

```python
# scatter plot with budget vs gross revenue


plt.scatter(x = df['budget'], y = df['gross'])
plt.title('budget vs gross earnings')
plt.xlabel('gross earnings')
plt.ylabel('budget')
plt.show()
```

```python
df.head()
```

```python
#plotting budget vs gross using seaborn

sns.regplot(x = 'budget', y = 'gross', data = df, scatter_kws = {"color" : "blue"}, line_kws = {"color" : "yellow"})
```

```python
#looking at correlation, have to specify only numeric values as the whole dataframe has string values as well

df.corr(numeric_only=True, method = 'pearson')
```

```python
#high correlation between budget and gross
```

```python
correlation_matrix = df.corr(numeric_only=True, method = 'pearson')
sns.heatmap(correlation_matrix, annot = True)
plt.title('correlation matrix for numeric features')
plt.xlabel('movie features')
plt.ylabel('movie features')
plt.show()
```

```python
df_numerize = df
for col_name in df_numerize.columns:
    if (df_numerize[col_name].dtype == 'str'):
        df_numerize[col_name] = df_numerize[col_name].astype('category')
        df_numerize[col_name] = df_numerize[col_name].cat.codes

df_numerize
```

```python
correlation_matrix = df_numerize.corr(numeric_only=True, method = 'pearson')
sns.heatmap(correlation_matrix, annot = True)
plt.title('correlation matrix for numeric features')
plt.xlabel('movie features')
plt.ylabel('movie features')
plt.show()
```
